```python
from collections import Counter

# Training data
text = "this is a good movie this is a nice movie"

# Count words
words = text.split()
counts = Counter(words)

# Vocabulary
vocabulary = set(words)

# Add an unseen word for testing
word = "bad"

# Total words and vocabulary size
total_words = len(words)
V = len(vocabulary)

# Laplace smoothing
probability = (counts[word] + 1) / (total_words + V)

print("Word:", word)
print("Count:", counts[word])
print("Probability after Laplace Smoothing:", probability)
```
