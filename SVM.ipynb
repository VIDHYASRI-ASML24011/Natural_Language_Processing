```python
# Text Classification using Support Vector Machine

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
```

```python
# Training documents
documents = [
    "India won the cricket match",
    "The player scored a century",
    "The movie was very entertaining",
    "The actor gave an excellent performance",
    "The government announced a new policy",
    "The election results were declared",
    "Artificial intelligence is growing rapidly",
    "The new smartphone has advanced technology"
]
```

```python
# Categories
labels = [
    "Sports",
    "Sports",
    "Entertainment",
    "Entertainment",
    "Politics",
    "Politics",
    "Technology",
    "Technology"
]
```

```python
# Convert text into TF-IDF vectors
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(documents)

# Create SVM classifier
model = SVC(kernel="linear")

# Train the model
model.fit(X, labels)

# Test sentences
test_documents = [
    "The cricket team won the tournament",
    "The new technology is very advanced",
    "The actor performed well in the movie",
    "The government introduced a new law"
]
```

```python
# Convert test text into TF-IDF vectors
X_test = vectorizer.transform(test_documents)

# Predict categories
predictions = model.predict(X_test)

# Display results
for text, prediction in zip(test_documents, predictions):
    print("Text:", text)
    print("Predicted Category:", prediction)
    print()
```
